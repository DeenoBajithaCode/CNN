```python
import numpy as np
import tensorflow
from  tensorflow.keras.layers import Conv2D, Activation, MaxPooling2D, Dense, Flatten
from tensorflow.keras import Sequential
```

```python
X_train = np.loadtxt('/content/drive/MyDrive/Colab Notebooks/CNN/input.csv', delimiter=',')
Y_train = np.loadtxt('/content/drive/MyDrive/Colab Notebooks/CNN/labels.csv', delimiter=',')

X_test = np.loadtxt('/content/drive/MyDrive/Colab Notebooks/CNN/input_test.csv',delimiter=',')
Y_test = np.loadtxt('/content/drive/MyDrive/Colab Notebooks/CNN/labels_test.csv',delimiter=',')
```

```python
print("Shape of X_train: ", X_train.shape)
print("Shape of Y_train: ", Y_train.shape)
print("Shape of X_test: ", X_test.shape)
print("Shape of Y_test: ", Y_test.shape)
```

```python
X_train = X_train.reshape(len(X_train),100,100,3)
Y_train = Y_train.reshape(len(Y_train),1)

X_test = X_test.reshape(len(X_test),100,100,3)
Y_test = Y_test.reshape(len(Y_test),1)
```

```python
print("Shape of X_train: ", X_train.shape)
print("Shape of Y_train: ", Y_train.shape)
print("Shape of X_test: ", X_test.shape)
print("Shape of Y_test: ", Y_test.shape)
```

```python
X_train[1,:]
```

```python
X_train = X_train/255.0
X_test = X_test/255.0
```

```python
X_train[1,:]
```

```python
import matplotlib.pyplot as plt
import random
idx= random.randint(0,len(X_train))
plt.imshow(X_train[idx,:])
```

```python
model = Sequential([
    Conv2D(32,(3,3),activation = 'relu',input_shape=(100,100,3)),
    MaxPooling2D((2,2)),
    Conv2D(32,(3,3),activation = 'relu'),
    MaxPooling2D((2,2)),
    Flatten(),
    #The above layers are for feature extraction
    #The following layers are for modelling (learning) with one hidden layer and a final output layer with a sigmoid activation function.
    Dense(64, activation='relu'),
    Dense(1, activation='sigmoid')
])
```

```python
model.compile(loss = 'binary_crossentropy' , optimizer ='adam', metrics=['accuracy'])
```

```python
model.fit(X_train, Y_train, epochs = 5, batch_size = 64)
```

```python
model.evaluate( X_test, Y_test, batch_size = 40, verbose=0)
```

```python
idx2 = random.randint (0,len(Y_test))
plt.imshow(X_test[idx2,:])
plt.show()

y_pred = model.predict(X_test[idx2,:].reshape(1,100,100,3))
print(y_pred)
```

```python
idx2 = random.randint (0,len(Y_test))
plt.imshow(X_test[idx2,:])
plt.show()
y_pred = model.predict(X_test[idx2,:].reshape(1,100,100,3))
if (y_pred > 0.5):
  pred = 'Cat'
else:
  pred = 'Dog'
print("The model prediction is a ", pred)
```
